## Puzzle

![](https://www.janestreet.com/puzzles/some_f_squares_2.png)

Place some non-overlapping "F"-shaped pentominos into the grid. Pentominos can be rotated, reflected, and enlarged. Place a 1 inside each square in an unenlarged pentomino. For a pentomino that has been enlarged by a factor of _N_ > 1, place an _N_ inside each square. The sum of the numbered squares inside each outlined region must be the same. A number outside the grid represents the sum of all numbered squares in the corresponding row or column of the completed grid.

Once completed, you can submit as your answer the product of the areas of the connected groups of orthogonally adjacent empty squares in the grid.
