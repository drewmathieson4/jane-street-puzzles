# Sum One, Somewhere

**Jane Street puzzle, April 2025** — [puzzle page](https://www.janestreet.com/puzzles/sum-one-somewhere-index/)

## Solution: .5306035754

## Puzzle

![](https://www.janestreet.com/puzzles/sum-one-somewhere.png)

For a fixed _p_, independently label the nodes of an infinite complete binary tree 0 with probability _p_, and 1 otherwise. For what _p_ is there exactly a 1/2 probability that there exists an infinite path down the tree that sums to **at most 1** (that is, all nodes visited, with the possible exception of one, will be labeled 0). Find this value of _p_ accurate to **10 decimal places**.

## Work

The rare puzzle that can be done on paper (with one computer software assist for solving a cubic polynomial).

### Deriving $S_0$

Let $p$ be the probability of a 0. Let $S_i$ be the probability that a binary tree contains an infinite path that sums to at most $i$.

Condition on the root's label. If the root is a 0 (probability $p$), we need at least one of the two child subtrees to contain a path summing to at most 0, which has chance $1-(1-S_0)^2$. If the root is a 1 (probability $1-p$), then we fail immediately.
$$S_0 = p\left(1-(1-S_0)^2\right) + (1-p)(0)$$

$$S_0 = p\left(2S_0 - S_0^2\right)$$

$$0 = (2p-1)S_0 - pS_0^2$$

$$0 = S_0\left(2p-1-pS_0\right)$$

So $S_0 = 0$ or $pS_0 = 2p-1$, giving

$$S_0 = \begin{cases} 0 & \text{if } p \le \tfrac{1}{2} \\[6pt] \dfrac{2p-1}{p} & \text{if } p > \tfrac{1}{2} \end{cases}$$

### Deriving $S_1$

With our expression for $S_0$ in terms of p, we can now derive the same expression for $S_1$.  If the root is a 0, we need to have one of the children satisfy $S_1$, and if the root is a 1, we need to have one of the children satisfy $S_0$.  Our expression for $S_1$ is therefore...

$$S_1 = p\left(1-(1-S_1)^2\right) + (1-p)\left(1-(1-S_0)^2\right)$$

We assume $p > \tfrac{1}{2}$, so $S_0 = \dfrac{2p-1}{p}$.

$$S_1 = p\left(2S_1 - S_1^2\right) + (1-p)\left(2S_0 - S_0^2\right)$$

$$S_1 = 2pS_1 - pS_1^2 + (1-p)\left(\frac{4p-2}{p} - \frac{4p^2-4p+1}{p^2}\right)$$

$$S_1 = 2pS_1 - pS_1^2 + (1-p)\left(\frac{4p^2-2p-4p^2+4p-1}{p^2}\right)$$

$$S_1 = 2pS_1 - pS_1^2 + (1-p)\left(\frac{2p-1}{p^2}\right)$$

$$S_1 = 2pS_1 - pS_1^2 + \frac{-2p^2+3p-1}{p^2}$$

Multiplying through by $p^2$:

$$p^2S_1 = 2p^3S_1 - p^3S_1^2 - 2p^2 + 3p - 1$$

$$0 = 2p^3S_1 - p^3S_1^2 - p^2S_1 - 2p^2 + 3p - 1$$

Let $S_1 = \tfrac{1}{2}$ as in the problem statement:

$$0 = p^3 - \tfrac{1}{4}p^3 - \tfrac{1}{2}p^2 - 2p^2 + 3p - 1$$

Multiplying through by 4:

$$0 = 4p^3 - p^3 - 2p^2 - 8p^2 + 12p - 4$$

$$0 = 3p^3 - 10p^2 + 12p - 4$$

Plugging the cubic into a solver gives

$$p = 0.5306035754\ldots$$
