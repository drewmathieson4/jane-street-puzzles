## Puzzle

The Artificial Automaton Athletics Association (Quad-A) is developing a Robot Baseball competition. Games consist of independent _at-bats_ where the batter maximizes expected score and the pitcher minimizes it.

An _at-bat_ is a series of _pitches_ with running counts of balls and strikes, both starting at zero. For each pitch, the pitcher secretly chooses to throw a ball or strike, and the batter secretly chooses to wait or swing. The results are:

* Ball + wait: balls increment by 1
* Strike + wait: strikes increment by 1
* Ball + swing: strikes increment by 1
* Strike + swing: with probability _p_ the batter hits a home run; with probability _1-p_ strikes increment by 1

An at-bat ends when:

* Balls reach 4: batter receives 1 point
* Strikes reach 3: batter receives 0 points
* Batter hits a home run: batter receives 4 points

Quad-A adjusts _p_ (probability a swung-at strike results in a home run) to maximize _q_, the probability of reaching a _full count_ (three balls and two strikes). Assuming both players use optimal mixed strategies and _p_ is chosen to maximize _q_, find this maximal _q_ to **ten decimal places**.
