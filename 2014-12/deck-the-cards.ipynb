```python
def print_letters(deck, values):
    letters = []
    for card in deck:
        value = values[card[:-1]]
        suit_value = suit_values[card[-1]]
        index = (value * 4 + suit_value) % 26
        letters.append(alphabet[index])
    # range(start, stop, step) counts 0, 13, 26, ... so each pass prints one row
    for row_start in range(0, len(letters), 13):
        row_of_letters = letters[row_start : row_start + 13]  # the next 13 letters
        # join glues the letters into one line with a space between each pair
        print(" ".join(row_of_letters))
```
