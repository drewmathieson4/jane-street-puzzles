## Puzzle

![](https://www.janestreet.com/puzzles/20190805_knight_moves_3.png)

Place the numbers 1, 2,…, _N_ (for some _N_) on a subset of the squares above, so that it is possible for a knight to move from 1 to _N_ via a series of legal knight's moves. Each number **inside** the grid represents the height of a building located at that square, and we can think of the knight as jumping from rooftop to rooftop on this series of incrementally taller buildings.

A number **outside** the grid indicates the first (i.e. smallest) number for which the knight was visible looking into the grid in the direction of that row or column. (As shown in the example.)

The answer to this puzzle is the **smallest achievable product** of the areas of the connected groups of empty squares in the completed grid.
