## Puzzle

![](https://www.janestreet.com/puzzles/20190903_block_party_3.png)

Fill each region with the digits 1 through _N_, where _N_ is the number of cells in the given region. For every cell in the grid, if _K_ denotes the number in that cell, then the nearest value of _K_ (looking only horizontally or vertically) must be located **exactly** _K_ cells away. (As in the example.) Some of the cells have already been filled in.

Once the grid is completed, take the largest "horizontally concatenated number" from each region and compute the sum of these values. **Enter this sum as your answer.** Good luck!
