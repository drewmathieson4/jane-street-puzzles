**Jane Street puzzle, November 2026** — [puzzle page](https://www.janestreet.com/puzzles/current-puzzle/)

# Solution - Antwerpsche Diamantkring

## Initial impression

The page contains just:

- a lone `Ø`,
- 75 strings of letters, each one alphabetised (`AAAABBIRRST`, `AAACDEGHHHILMPRS`, ... `SW`, `T`, `V`),
- 19 emoji: 👸 🧭 🐷 ⚽ 🎮 🖼 🧐 ⚾ 💻 🐉 🕹 🏛 🎶 ⭐ 🕓 ⛵ 🏜 🪷 ⚠
- and an enumeration for the answer: `(11 12)`.

Some strings unscramble cleanly or are multi-anagrams, including some, like INSTABILITY and CAPABILITY, that can't simply be random.  So we need to anagram the letters.  

There are 19 emojis and 76 sets of alphabetized letters, so I suspect each emoji clues 4 anagrams.

## The mechanic
The obvious emojis like the compass, the clock, and the boat have diamond shaped sets of 4 associated with them (north, east, south, west), (three, six, nine, twelve), and (port, bow, stern, starboard).

Pressure makes diamonds, hence *Pressure Points* is a clue towards this mechanic, where each emoji represents a diamond of things.  Additionally, in every set of four:

- one entry has had a letter **added**,
- one entry has had a letter **removed**,
- the other two are plain anagrams.

`Ø` must therefore be a single letter, where the letter was then removed.

## Extraction

Reading the changed letters down the table, in emoji order:

- **Added:** N E W E S T R O U G H E X C H A N G E → **NEWEST ROUGH EXCHANGE**
- **Removed:** L E U V E N G H E N T B R U S S E L S → **LEUVEN GHENT BRUSSELS**

Leuven, Ghent and Brussels are three Belgian cities, so the puzzle's own trick has been applied to its final diamond: we need to find the fourth Belgian city, which geographically is solved by taking **Antwerp**, unsurprisingly the diamond capital of the world.

Googling "Antwerp newest rough exchange" gives us a diamond exchange founded in 1929 specifically for rough diamonds, called the **Antwerpsche Diamantkring**, which matches the `(11 12)` clue at the end.

### Final answer

**Antwerpsche Diamantkring**
