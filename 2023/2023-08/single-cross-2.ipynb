## Puzzle

Consider 3-space (i.e. **R**³) partitioned into a grid of unit cubes with faces defined by the planes of all points with at least one integer coordinate. For a fixed positive real number _D_, a random line segment of length _D_ (chosen uniformly in location and orientation) is placed in this cubic lattice.

What length _D_ maximizes the probability that the endpoints of the segment lie in orthogonally adjacent unit cubes (that is, the segment crosses exactly one integer-coordinate plane), and what is this maximal probability? Give your answer as a **comma-separated pair of values to 10 significant places** (e.g. "1.234567891,0.2468135792").
