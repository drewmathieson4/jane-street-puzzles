## Puzzle

![](https://www.janestreet.com/puzzles/circle_time_01.PNG)

Call a "ring" of circles a collection of six circles of equal radius, say r, whose centers lie on the six vertices of a regular hexagon with side length 2r. This makes each circle tangent to its two neighbors, and we can call the center of the regular hexagon the "center" of the ring of circles. If we are given a circle _C_, what is the maximum proportion of the area of that circle we can cover with rings of circles entirely contained within _C_ that all are mutually disjoint and share the same center?
