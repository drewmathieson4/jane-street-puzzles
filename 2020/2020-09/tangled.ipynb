## Puzzle

While on vacation in Terniquaternaria, Decimus experienced the country's traditional folkloric dance called the "Yawn-ock". The people there are rational and dance by strict rules. Four dancers hold one end of two long pieces of rope and never let go, initially positioned as:

```
A           D
 \_________/
  _________
 /          \
B            C
```

Two types of moves are allowed:

* "The entire group rotates 90 degrees counterclockwise (ie the dancer in A moves to B, dancer in B moves to C, etc)". This is called "Rotate" (R).
* "Dancers in C and D swap places, with D holding the rope up while C is passing under." This is called "Swap" (S).

A dance master calls out random R/S steps for dancers to execute. Afterward, a crowd participant attempts to untangle the ropes using the same moves. Success earns them King/Queen for the day.

Decimus noted this sequence executed by dancers:

```
SRSRRSSRSRSSRSSRRSSRSSSSSRSSRSSRSRS
SRSSRSSSSSSSSRSSRSSSSSRSSRSSRRSSRSS
SSSRSSRSSRSSSSSSSSSSSSSSSSSRSSRSSRS
```

**What sequence would untangle the ropes?**
