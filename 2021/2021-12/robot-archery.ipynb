## Puzzle

After a grueling year filled with a variety of robot sporting events, we arrive at the final competition: **Robot Archery**. Four robots qualified for the finals with these seeds:

| Robot | Seed |
|-------|------|
| Aaron | 1 |
| Barron | 2 |
| Caren | 3 |
| Darrin | 4 |

## Rules

"When it is a given robot's turn, they shoot a single arrow. If it is closer to the center of the target than _all_ previous arrows by all players, that robot remains in the tournament, going to the back of the queue to await their next turn. Otherwise that robot is eliminated immediately."

The robots take turns in seed order, beginning with Aaron. The last remaining robot wins.

## Question

To ten decimal places, what is the probability that **Darrin** will be this year's winner?

**Note:** "Each robot is equally skilled. Which is to say: for any region _R_ on the target with nonzero area, the robots all have the same positive probability of landing an arrow within _R_ on any given shot."
