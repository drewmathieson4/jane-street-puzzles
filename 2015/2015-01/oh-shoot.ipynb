# Oh SHOOT! — Jane Street, January 2015

https://www.janestreet.com/puzzles/oh-shoot-index/

![Oh SHOOT!](https://www.janestreet.com/puzzles/OH_SHOOT.jpg)

## Solution: None/Failed

## Attempts

I'll admit to spending a while on this one without getting anywhere.  I noticed the chess notation right away, peppered with unconventional things like the simple xa5, z5, Ki2.  I noticed that the addition of z, i, 0, and 9 meant we were likely playing on a 10x10 boards, with the z being to the left of a and zero being below 1 (I played around briefly with a 26x10 board).

Next was to interpret the "Oh Shoot" line.  Ultimately, I came upon a chess variant called rifle chess, in which pieces don't move when they capture (they shoot the piece from a distance).  I thought this was for sure the answer, as it elegantly explains the xa5 notation (no piece moves to capture).  I set about recreating a game of rifle chess that results in these moves, which proved immensely hard.  Moreover, the move listed as 1 starts with a queen capture, which is impossible even in rifle chess.  I played around a bit and then eventually gave up and looked up the solution.

## The solution

As it turns out, the solution was a chess game on a 10x10 board, but that board was the class Shoots and Ladders board!  Never in a million years would I have figured it out.
