# Ticker Treat

**Jane Street puzzle, March 2015** — [puzzle page](https://www.janestreet.com/puzzles/ticker-treat-index/)

## Puzzle

The answer to this month's puzzle is a letter of the alphabet.

1. They cause auroras (5,5)
2. Cross where it's not too deep (4)
3. Cricket owner (2&1)
4. Ancient box-opener? (7)
5. Wheel part, to a shark? (3)

## Solution 

**S**


## Solution

### Initial clues

1. They cause auroras (5,5) -- SOLAR WINDS
2. Cross where it's not too deep (4) -- WADE
3. Cricket owner (2&1) -- AT&T
4. Ancient box-opener? (7) -- PANDORA
5. Wheel part, to a shark? (3) -- ACE


Number 3 refers to cricket wireless.
Number 5 uses the fact that a wheel in poker (poker shark) is an A-2-3-4-5 straight.  Jane street seemingly loves poker.
