## Puzzle

A particle is located at the center of the square marked ①, and wishes to travel to the center of the square marked ②.

The particle has a rope that it can toss to any of the posts (●'s) that it can see, provided there are no other posts in the way. (I.e., no intermediate collinear posts.)

Once connected to a post, the particle can freely swing (rotate) within the boundaries of the board, but may only come to rest at the center of a post-less square. The particle may not pass through a post. After the swing, though, the swinging post is removed from the board.

Each swing costs the particle $D^{-2}$, where D is the straight-line distance of the length of rope used for that particular swing.

(Note that the other posts on the board may interfere with the path of the swinging rope, though you should think of the posts as having zero width. See the Example, below. Furthermore, the rope is allowed to cross itself.)

**Your goal for this puzzle is to get from ① to ② for as small a cost as you can.**

To enter, send in your total cost (rounded to the nearest .0001) and your swing (move) list. For notational purposes, each swing should be listed as an ordered pair (P, S), where P is the swinging Post's location, and S is the Stopping point for the swing. (See the Example, below.)

The puzzle board is a 20-row (1-20) by 20-column (a-t) grid with posts (●) scattered through it, the start ① at a1, and the goal ② at t20. The board itself is only available as an image inside the source PDF, not as a standalone image file, so it isn't reproduced here. _See the puzzle PDF for the exact board layout: [May18_Puzzle.pdf](https://www.janestreet.com/static/puzzles/May18_Puzzle.pdf)_
