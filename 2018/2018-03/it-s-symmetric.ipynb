## Puzzle

![](https://www.janestreet.com/puzzles/20180301_symmetric.png)

The grid below can be partitioned into several regions, each of which is symmetric in some way (be it rotation or reflection). A cell with a number indicates "the length of the shortest walk, within that region, to an 'opposite' cell" (a cell that could potentially be moved to the numbered cell's location under a non-trivial rotation or reflection of the plane).

Examples provided: In a 1-by-N region, all cells could be labeled 0, since reflection across the longer mid-line takes every cell to itself. In the "S"-shaped Tetris piece, cells could be labeled 3, 1, 1, 3.

**The answer to this puzzle is the sum of the cubes of the areas of the regions in the completed grid.**
